```python
#Excercise
#Train SVM classifier using sklearn digit dataset
import pandas as pd
from sklearn.datasets import load_digits
digits=load_digits()
```

```python
dir(digits)
#column
```

```python
digits.target_names
```

```python
df=pd.DataFrame(digits.data,digits.target)
df.head()
```

```python
df['target']=digits.target
df.head()
#adding one more column target and adding this to the previousdataframe
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(df.drop('target',axis='columns'),df.target,test_size=0.2)
```

```python
from sklearn.svm import SVC
rbf_model =SVC(kernel='rbf')
#If your data is linearly separable, use the linear kernel (faster, simpler).
#If your data has complex patterns, use RBF kernel (more powerful).
```

```python
rbf_model.fit(X_train,Y_train)
```

```python
rbf_model.score(X_test,Y_test)
```

```python
#using linear kernel
linear_model=SVC(kernel='linear')
linear_model.fit(X_train,Y_train)
```

```python
linear_model.score(X_test,Y_test)
```
