```python
#To decide which one model is perfect for the dataset
#to check the models performance
#we divide the dataset into fold
#example like 100 samples in a datset
# we divide the dataset into 20 folds and in first iteration i train the 4 fold and first one for test and in second iteration i test the second fold and remaining for training and do so and average score is taken

from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
import numpy as np
from sklearn.datasets import load_digits
digits = load_digits()
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(digits.data,digits.target,test_size=0.3)
```

```python
lr=LogisticRegression()
lr.fit(X_train,y_train)
lr.score(X_test,y_test)
```

```python
svm=SVC()
svm.fit(X_train,y_train)
svm.score(X_test,y_test)
```

```python
rf=RandomForestClassifier()
rf.fit(X_train,y_train)
rf.score(X_test,y_test)
```

```python
#when we re-execute the X_train and x_test and the sample we are putting into model changes so the accuracy also changes
#so train test split will not always give accurate performance so we use k fold
from sklearn.model_selection import KFold
kf=KFold(n_splits=3)
kf
#spliting the datasets into k fold
```

```python
for train_index ,test_index in kf.split([1,2,3,4,5,6,7,8,9]):
      print(train_index,test_index)
# it divide the datset into 3 fold because earlier we have mentioned it to divide it to divide it into 3 folds and the data we have given here i 1 to 9
```

```python
def get_score(model,X_train,X_test,y_train,y_test):
    model.fit(X_train,y_train)
    return model.score(X_test,y_test)
```

```python
get_score(LogisticRegression(),X_train,X_test,y_train,y_test)
#This function trains a machine learning model on training data and evaluates its accuracy on test data.
```

```python
get_score(SVC(),X_train,X_test,y_train,y_test)
```

```python
from sklearn.model_selection import StratifiedKFold
folds=StratifiedKFold(n_splits=3)
#StratifiedKFold is a way to split data into training and testing sets while making sure that the proportion of each class (category) remains the same in every split.
```

```python
#Three empty lists are created to store model scores:
score_l=[]
score_svm=[]
score_rf=[]
for train_index,test_index in kf.split(digits.data):
    X_train,X_test,y_train,y_test=digits.data[train_index],digits.data[test_index],digits.target[train_index],digits.target[test_index]
    get_score(LogisticRegression(),X_train,X_test,y_train,y_test)
    get_score(SVC(),X_train,X_test,y_train,y_test)
    get_score(RandomForestClassifier(),X_train,X_test,y_train,y_test)

print("Logistic Regression Scores:", score_l)
print("SVM Scores:", score_svm)
print("Random Forest Scores:", score_rf)
```
